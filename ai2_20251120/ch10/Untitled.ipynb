```python
def toOtherType(data):
    result = []
    for v in data:
        result.append(int(v))
    return   result
myList = ["12","35","77"]
print(myList)
print(toOtherType(myList))
```

```python
def toOtherType(data):
    result = []
    for v in data:
        result.append(str(v))
    return   result
myList = [12,35,77]
print(myList)
print(toOtherType(myList))
```

```python
def toOtherType(data):
    result = []
    for v in data:
        if v % 2 == 0:
            result.append(v)
    return   result
myList = [12,35,77,66,88,91]
print(myList)
print(toOtherType(myList))
```

```python
def myMap(func,data):
    result = []
    for v in data:
       result.append(func(v))
    return result
myList = ["12","66","25","89"]
data1 = myMap(int,myList)
print(data1)

print(int("12"))
```

```python
def myFilter(func,data):
    result = []
    for v in data:
        if func(v):
           result.append(v)
    return result
def testFunc(v):
    if v % 2 == 0:
        return True
    else:
        return False
    
myList = [12,6,9,3,15,8]    
print(myFilter(testFunc,myList))    
```

## Map

```python
myList = ["12","66","25","89"]
mapObj =  map(int,myList)
mapObj = list(mapObj)
print(mapObj)
```

```python
list3 = [1,2,3,'',4,5,'']
def emptyToZero(v):
    return v if v != '' else 0
mapObj = map(emptyToZero,list3)
mapObj  = list(mapObj)
print(mapObj)
```

```python
mapObj = map(lambda x: x if x != "" else 0,list3)
print(list(mapObj))
```

```python
list4 = ["M","W","M","M","W"]
#M轉換成0 W轉換成1
objMap = map(lambda x:0 if x == "M" else 1,list4)
print(list(objMap))
```

```python
list4 = ["M","W","M","M","W"]
myDict = {"M":0,"W":1}
objMap = map(lambda x : myDict[x],list4)
print(list(objMap))
```

## Filter

```python
a = [1,2,3,4,5,6]
filterObj = filter(lambda x : x%2 == 0,a)
print(list(filterObj))
```

```python
list3 = [1,2,3,'',4,5,'']
#不是''幫我輸出成list
filtetObj = filter(lambda x : x != "",list3)
print(list(filtetObj))
```

## 例外語法
try:
    保護的程式區段
    
except 例外1 :
    例外1發生時執行的程式
    
except :
    其他例外發生時執行的程式
    #未指定例外型態，捕捉所有例外物件
    
else:
    若try程式區段沒產生例外執行的程式
    
finally:
    不管有沒有發生例外都會執行的程式

```python
num1 = 10
num2 = 1
print("Start")
try:
    print(num1 / (num2-1))
except:
    print("發生錯誤")
print("End")
```

```python
num1 = 10
num2 = 1
num3 = [1,2,3,4]
print("Start")
try:
    print(num3[1])
    print(num1 / num2)
except:
    print("發生錯誤")
print("End")
```

```python
num1 = 10
num2 = 1
num3 = [1,2,3,4]
print("Start")

try:
    print(num3[9])
    print(num1 / num2)
except IndexError :
    print("IndexError")
except:
    print("發生錯誤")
print("End")
```

```python
list1 = [1,2,3]
try:
    print(list1.pop())
    print(list1.pop())
    print(list1.pop())
    print(list1.pop())#顯示List為空
except:
    print("List為空")
```

## 使用raise拋例外

```python
def test(v1):
    if v1 < 0 :
        raise OverflowError("不可小於0")
    else:
        print(v1)
try:
    test(5)        
    test(-7)
except OverflowError as ex:
    print("OverflowError:",ex)
    
 
```
